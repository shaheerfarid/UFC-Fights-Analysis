```python
import pandas as pd
```

```python
tables = {}
names = ["ufc_event_details", "ufc_fight_details", "ufc_fight_results", "ufc_fight_stats", "ufc_fighter_details", "ufc_fighter_tott"]
for name in names:
    tables[name] = pd.read_csv(f"https://raw.githubusercontent.com/Greco1899/scrape_ufc_stats/main/{name}.csv")
```

```python
for name, df in tables.items():
    print(name, df.shape)
```
